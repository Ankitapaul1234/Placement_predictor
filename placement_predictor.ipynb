```python
import numpy as np
import pandas as pd
```

```python
#steps

#1.process + EDA + Feature selection
#2.Scale the values
#3.Train test split
#4.Train the model
#5.Evaluate the model/model selection
#6.Deploy the model
```

```python
df = pd.read_csv('/content/placement.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df = df.iloc[:,1:]
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
# EDA
plt.scatter(df['cgpa'],df['iq'],c=df['placement'])
```

```python
#Feature selection  ==> dependent and independent col selection
X = df.iloc[:,0:2]   #X-> independent col(iq,cgpa) :,0:2=>all row and 0 to 1 col
Y = df.iloc[:,-1]    #Y->dependent col(placement)  :,-1 => all row and last col
```

```python
X
```

```python
Y
```

```python
#train and test
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.1)  #test_size=0.1 means for testing=>10% dataset and for train=>90%
```

```python
X_train
```

```python
X_test
```

```python
X_test.shape
```

```python
Y_train
```

```python
#scaling
from sklearn.preprocessing import StandardScaler  #StandardScaler is a class
```

```python
scaler = StandardScaler() #scaler is a obj of StandardScaler class
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_train
```

```python
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression()
```

```python
#train the model using LogisticRegression
clf.fit(X_train,Y_train)
```

```python
clf.predict(X_test)
```

```python
#Evaluate the model/model selection
y_pred = clf.predict(X_test)
```

```python
Y_test
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test,y_pred)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X_train,Y_train.values,clf=clf,legend=2)
```
